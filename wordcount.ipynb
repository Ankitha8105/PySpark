```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import explode, split, col
from pyspark.sql.types import StringType
```

```python
spark = SparkSession.builder \
    .appName("PySpark Word Count from CSV") \
    .getOrCreate()

input_file = "description.csv" 
df = spark.read.option("header", "true").csv(input_file)

df.show()

words_df = df.select(explode(split(col("text"), " ")).alias("word"))

words_df = words_df.filter(words_df["word"] != "")

word_count_df = words_df.groupBy("word").count().sort(col("count").desc())

word_count_df.show()

word_count_df.write.option("header", "true").csv("output_wordcount.csv")

spark.stop()
```
