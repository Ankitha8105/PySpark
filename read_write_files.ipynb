```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.appName('read_txt_file').getOrCreate()

input_file = "sampleData.txt"
file = spark.read.text(input_file)

output_text = "output_data_text"
file.write.mode("overwrite").text(output_text)

file.show()

spark.stop
```

### Read Write the Csv File using PySpark

```python
spark = SparkSession.builder.appName("read_csv").getOrCreate()

input_file = "test1.csv"
csv_df = spark.read.option("header", "true").csv(input_file)

output_csv = "output_data_csv"
csv_df.write.option("header", "true").mode("overwrite").csv(output_csv)

file.show()

spark.stop
```

### Read and Write the JSON file 

```python
spark = SparkSession.builder.appName("read_write_json").getOrCreate()

input_file = "description.json"
json_df = spark.read.json(input_file)

json_df.show()

output_json = "output_data_json"
json_df.write.mode("overwrite").json(output_json)

spark.stop
```

```python
spark = SparkSession.builder.appName("ReadWriteParquet").getOrCreate()

input_parquet = "titanic.parquet"
parquet_df = spark.read.parquet(input_parquet)

print("Data read from Parquet:")
parquet_df.show()

output_par ="output_data_parquet"
parquet_df.write.mode("overwrite").parquet(output_par)

spark.stop
```
